# **Seatwork 1.4: Lists, Dictionaries, and Tuples**



```python
fruits = ["Saging, saging", "mangga, mangga", "papaya...papayaa"]
print("Original list:", fruits)
print("First Fruit  : ", fruits[0])
print("Second Fruit : ", fruits[1])
print("Third Fruit  : ", fruits[2])
fruits.append("banana, banana")
fruits.insert(4, "fruit salad, fruit salad")
fruits.remove("papaya...papayaa")
popped = fruits.pop()
print("After adding fruits", fruits)
print("Evictus      : ", popped)

if "Saging, saging" in fruits:
  print("\nEverybody sing!!!!")


```

```python
my_tuple = (1, 2, True, "a string", (3, 4), [5, 6], None)
print(my_tuple)
my_list = [1, 2, True, "a string", (3, 4), [5, 6], None]
print(my_list)
```

```python
num_dictionary = {1: "one", 2: "two", 3: "three", 4: "four", 5: "five", 6: "six", 7: "seven", 8: "eight", 9: "nine"}
item_2 = num_dictionary[3]
print(item_2)
```
