# Part 2: Conditional Execution

Exercise 1: Rewrite your pay computation to give the employee 1.5 times the hourly rate for hours worked above 40 hours.



```python
hour = input("Enter working hours: ")
rate = input("Enter Rate: ")
h = float(hour)
r = float(rate)

if h > 40:
  pay = (40 * r) + (h - 40) * (r * 1.5)
else:
  pay = h * r

print("Gross pay: ₱", pay)

```

Exercise 2: Rewrite your pay program using try and except so that your prog

```python
hour = input("Enter working hours: ")
rate = input("Enter Rate: ")

try:
    h = float(hour)
    r = float(rate)

    if h > 40:
        pay = (40 * r) + (h - 40) * (r * 1.5)
    else:
        pay = h * r

    print("Gross pay: ₱", pay)

except:
    print("Error, please enter numeric input")
```
