**Exercise 1:** Write a program which repeatedly reads integers until the user enters “done”. Once “done” is entered, print out the total, count, and average of the integers. If the user enters anything other than a integers, detect their mistake using try and except and print an error message and skip to the next integers.

````python
total = 0
sum = 0
average = 0
print("Enter a number as many as you like and enter done once you finished")
count = input("Enter here: ")
while  count != "done":
  try:
    num = int(count)
    sum = sum + num
    total = total + 1
    average = sum / total
  except:
    print("Invalid input. Please34 enter an integer only")
  count = input("Enter here: ")


print("Number of input: ", total)
print("Sum of input: ", sum)
print("Average of input: ", average)
````
